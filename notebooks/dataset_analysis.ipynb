# Dataset Analysis
This notebook covers dataset analysis including class counts, sample images, and imbalance checks.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
import os
import glob
import matplotlib.image as mpimg

df = pd.read_csv('../data/farm_dataset.csv')
df.head()
```

## Class Counts

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='crop')
plt.title('Crop Class Counts')
plt.xticks(rotation=45)
plt.show()

print("Class counts:")
print(df['crop'].value_counts())
```

## Sample Images

```python
image_paths = glob.glob('../data/**/*.jpg', recursive=True)
if not image_paths:
    image_paths = glob.glob('../data/**/*.png', recursive=True)

if image_paths:
    plt.figure(figsize=(12, 8))
    for i, img_path in enumerate(image_paths[:6]):
        plt.subplot(2, 3, i + 1)
        img = mpimg.imread(img_path)
        plt.imshow(img)
        plt.title(os.path.basename(img_path))
        plt.axis('off')
    plt.show()
else:
    print("No sample images found in ../data/.")
```

## Imbalance Check

```python
class_counts = df['crop'].value_counts()
min_class = class_counts.min()
max_class = class_counts.max()
imbalance_ratio = max_class / min_class

print(f"Most frequent class count: {max_class}")
print(f"Least frequent class count: {min_class}")
print(f"Imbalance ratio: {imbalance_ratio:.2f}")

if imbalance_ratio > 1.5:
    print("Warning: Class imbalance detected!")
else:
    print("Dataset seems relatively balanced.")
```

# Data Augmentation
Using tf.keras.Sequential with RandomFlip, RandomRotation, RandomZoom, and RandomContrast.

```python
from tensorflow.keras import layers

data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomContrast(0.2)
])

print("Data augmentation pipeline created:", data_augmentation)
```
